### Part A - Prepare the offline dataset
The built-in scikit-learn `load_digits()` dataset supplies 1,797 small digit images. No internet downloads are needed after the Python libraries have been installed.

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import load_digits
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, ConfusionMatrixDisplay

SEED = 42
np.random.seed(SEED)
digits = load_digits()
X = digits.data.astype(float) / 16.0
y = digits.target.astype(int)
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.25, random_state=SEED, stratify=y
)

def build_model(train_features, train_labels):
    model = LogisticRegression(max_iter=600, random_state=SEED)
    model.fit(train_features, train_labels)
    return model

model = build_model(X_train, y_train)
print('Dataset:', X.shape, '| Train:', len(X_train), '| Test:', len(X_test))
print('Clean test accuracy:', round(accuracy_score(y_test, model.predict(X_test)), 4))
```

### Part B - Inspect examples and model performance

```python
fig, axes = plt.subplots(2, 5, figsize=(9, 4))
for ax, idx in zip(axes.ravel(), range(10)):
    ax.imshow(X_test[idx].reshape(8, 8), cmap='gray_r', vmin=0, vmax=1)
    ax.set_title(f'True {y_test[idx]} | Pred {model.predict(X_test[idx:idx+1])[0]}', fontsize=9)
    ax.axis('off')
plt.tight_layout(); plt.show()
ConfusionMatrixDisplay.from_estimator(model, X_test, y_test, cmap='Blues')
plt.title('Baseline confusion matrix'); plt.show()
```

**Reflection:** Why does good clean-test accuracy not establish security against manipulated inputs?

Clean-test accuracy only measures a model's performance on ordinary, unperturbed samples drawn from a benign target distribution[cite: 1]. A model can achieve high clean accuracy while maintaining fragile high-dimensional decision boundaries[cite: 1]. Adversarial evaluation tests worst-case performance under an active opponent who intentionally calculates perturbations designed to cross those decision boundaries—a risk that clean-test metrics cannot evaluate[cite: 1].
